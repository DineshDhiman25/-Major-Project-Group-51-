```python
import pandas as pd
import os

# Read CSV
df = pd.read_csv("translated_posts.csv")

# Create output folder if it doesn't exist
os.makedirs("split_csv", exist_ok=True)

# Define chunk size
chunk_size = 178

for i in range(30):
    start = i * chunk_size
    end = start + chunk_size
    df_chunk = df.iloc[start:end]

    # Keep only relevant columns
    df_chunk = df_chunk[['Channel ID', 'Post ID', 'Post']].copy()

    # Add serial number starting from 1 for each file
    df_chunk.insert(0, 'Sr. no.', range(1, len(df_chunk) + 1))

    # Save each chunk to CSV
    df_chunk.to_csv(f"split_csv/post{i+1}.csv", index=False)

print("✅ Done! 30 split CSV files created successfully in 'split_csv' folder.")
```

## Segregating Dataset

```python
import pandas as pd
import re

df = pd.read_csv("/Major_Grp51/Aditya/dataset/telegram_posts_1002843824.csv")

# ── English keywords ──────────────────────────────────────────────────────────
english_keywords = [
    "kill", "killed", "killing", "murder", "murdered",
    "torture", "tortured", "torturing",
    "rape", "raped", "sexual assault", "sexual violence",
    "execution", "executed", "hang", "hanged", "hanging",
    "flogging", "lashing", "whipping",
    "arrest", "detained", "imprisonment", "imprisoned",
    "crackdown", "suppression", "violence", "brutality",
    "death sentence", "death penalty",
    "forced disappearance",
    "human rights violation", "abuse"
]

# ── Persian / Farsi keywords ──────────────────────────────────────────────────
persian_keywords = [
    # kill / murder
    "کشتن", "کشته", "کشت", "قتل",
    # Militiary
    "نیروهای مسلح", "نیروهای دفاعی", "سربازان", "ارتش", "سربازان", "هنگ", "گردان", "شبه‌نظامیان",
    # torture
    "شکنجه",
    # rape / sexual violence
    "تجاوز", "تجاوز جنسی", "آزار جنسی",
    # execution / hanging
    "اعدام", "دار زدن", "به دار آویختن",
    # # arrest / detention / prison
    "دستگیری", "بازداشت", "زندان", "حبس", "زندانی",
    # crackdown / violence / massacre
    "سرکوب", "خشونت", "کشتار",
]

# ── Patterns ──────────────────────────────────────────────────────────────────
# Word-boundary (\b) works for Latin; Persian uses substring match
latin_pat = re.compile(
    r'\b(' + '|'.join(re.escape(k) for k in english_keywords) + r')\b',
    re.IGNORECASE
)
persian_pat = re.compile(
    '(' + '|'.join(re.escape(k) for k in persian_keywords) + ')'
)

def is_hrv(text):
    if not isinstance(text, str):
        return False
    return bool(latin_pat.search(text)) or bool(persian_pat.search(text))

def matched_keywords(text):
    if not isinstance(text, str):
        return ""
    hits = (
        [m.group() for m in latin_pat.finditer(text)] +
        [m.group() for m in persian_pat.finditer(text)]
    )
    return ", ".join(sorted(set(hits)))

# ── Filter & save ─────────────────────────────────────────────────────────────
mask = df["text_preview"].apply(is_hrv)
hrv_df = df[mask].copy()
hrv_df["matched_keywords"] = hrv_df["text_preview"].apply(matched_keywords)

hrv_df.to_csv("/Major_Grp51/Aditya/dataset/Iran/New/iran_new_posts.csv", index=False)
print(f"Total posts     : {len(df)}")
print(f"HRV posts found : {len(hrv_df)}")
print(f"Saved")
```

```python
import pandas as pd
import re

df = pd.read_csv("/Major_Grp51/Aditya/dataset/Venezuela/v_all.csv")

# ── Spanish keywords (HRV-focused) ─────────────────────────────────────────────
spanish_keywords = [
    # killing / death
    "matar", "mató", "muerto", "muertos", "asesinar", "asesinato",
    "homicidio", "ejecución", "ejecutado", "ejecuciones",
    # torture / abuse
    "tortura", "torturado", "maltrato", "abuso",
    # sexual violence
    "violación", "violado", "violencia sexual", "abuso sexual",
    # arrest / detention
    "detención", "detenido", "arresto", "encarcelado",
    "preso", "prisión", "reclusión",
    # repression / crackdown
    "represión", "represivo", "represalia",
    "brutalidad", "violencia", "uso excesivo de la fuerza",
    # disappearances
    "desaparición forzada", "desaparecido", "desaparecidos",
    # protests
    "protesta", "manifestación", "manifestantes",
    # human rights
    "derechos humanos", "violación de derechos",
    # death penalty
    "pena de muerte"
]

# ── Pattern ────────────────────────────────────────────────────────────────────
spanish_pat = re.compile(
    r'\b(' + '|'.join(re.escape(k) for k in spanish_keywords) + r')\b',
    re.IGNORECASE
)

def is_hrv(text):
    if not isinstance(text, str):
        return False
    return bool(spanish_pat.search(text))

def matched_keywords(text):
    if not isinstance(text, str):
        return ""
    hits = [m.group() for m in spanish_pat.finditer(text)]
    return ", ".join(sorted(set(hits)))

# ── Filter & save ──────────────────────────────────────────────────────────────
mask = df["text_preview"].apply(is_hrv)
hrv_df = df[mask].copy()
hrv_df["matched_keywords"] = df.loc[mask, "text_preview"].apply(matched_keywords)

hrv_df.to_csv(
    "/Major_Grp51/Aditya/dataset/Venezuela/venezuela_kill_posts.csv",
    index=False
)

print(f"Total posts     : {len(df)}")
print(f"HRV posts found : {len(hrv_df)}")
print("Saved")
```
